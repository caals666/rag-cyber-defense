# Dataset Inspection: CVE, CWE, CAPEC, ATT&CK

Reads the four cybersecurity knowledge graph datasets and displays their
headers, shapes, and sample rows before loading them into Neo4j.

```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 60)
```

## File paths

Update these if your CSVs are named or located differently.

```python
FILES = {
    "CVE":    "cve_dataset.csv",
    "CWE":    "cwe_dataset.csv",
    "CAPEC":  "capec_dataset.csv",
    "ATTACK": "attack_dataset.csv",
}

N_ROWS = 5
```

## CVE Dataset

```python
cve_df = pd.read_csv(FILES["CVE"])
print(f"Rows: {len(cve_df)}   Columns: {len(cve_df.columns)}")
print("\nColumn headers & dtypes:")
print(cve_df.dtypes)
cve_df.head(N_ROWS)
```

## CWE Dataset

```python
cwe_df = pd.read_csv(FILES["CWE"])
print(f"Rows: {len(cwe_df)}   Columns: {len(cwe_df.columns)}")
print("\nColumn headers & dtypes:")
print(cwe_df.dtypes)
cwe_df.head(N_ROWS)
```

## CAPEC Dataset

```python
capec_df = pd.read_csv(FILES["CAPEC"])
print(f"Rows: {len(capec_df)}   Columns: {len(capec_df.columns)}")
print("\nColumn headers & dtypes:")
print(capec_df.dtypes)
capec_df.head(N_ROWS)
```

## ATT&CK Dataset

```python
attack_df = pd.read_csv(FILES["ATTACK"])
print(f"Rows: {len(attack_df)}   Columns: {len(attack_df.columns)}")
print("\nColumn headers & dtypes:")
print(attack_df.dtypes)
attack_df.head(N_ROWS)
```

## Summary

Quick overview of all four datasets side by side.

```python
summary = pd.DataFrame({
    "dataset": list(FILES.keys()),
    "rows": [len(cve_df), len(cwe_df), len(capec_df), len(attack_df)],
    "columns": [len(cve_df.columns), len(cwe_df.columns), len(capec_df.columns), len(attack_df.columns)],
})
summary
```
